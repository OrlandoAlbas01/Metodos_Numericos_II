#Integración de Romberg.

Aproximar la integral:

$$\int_0^\pi\sin x\,dx$$

cuyo valor exacto es $2$

```python
import numpy as np
```

```python
def Romberg(a,b,n,f):
  R = np.zeros((n,n))

  h = b-a
  R[0, 0]=(h/2) * (f(a) + f(b))

  for i in range(1,n):
   k = np.arange(1,2**(i-1) + 1)
   suma = np.sum(f(a + (k-0.5)*h))
   R[i,0]=0.5*(R[i-1,0] + h*suma)

   for j in range(1,i+1):
     R[i,j]=R[i,j-1]+(R[i,j-1] - R[i-1,j-1])/(4**j-1)

  h=h/2

  return R
```
