#Regla compuesta de Simpson

$$\int_a^b f(x)\dx \aproxx \frac{h}{3}\left[f(a)+2\sum_{j=1}^{(n=2)-1}f(x_{2j})+4\sum_{j=1}^{n/2}f(x_{2j-1})+f(b)\right]$$

```python
def Simpson_compuesta(f, a, b, n):
  if n%2 != 0 or n <= 0:
    raise ValueError("n debe ser un entero positívo par")

  h = (b - a)/n
  XI0 = f(a) + f(b)
  XI1 = 0
  XI2 = 0
  for i in range (1, n):
    X = a +i*h
    if i%2 == 0:
      XI2 += f(X)
    else:
      XI1 += f(X)
  XI = h*(XI0 + XI2*2 + XI1*4)/3
  return XI
```

```python
import math

def f(x):
  return math.exp(2*x)*math.sin(3*x)

print(Simpson_compuesta(f, 0, 2, 52))
```
