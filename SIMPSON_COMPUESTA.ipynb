#Regla compuesta de Simpson.

Dado el ejercicio de la tarea anterior:

$$\int_0^2 e^{2x}\sin(3x)\,dx$$

usando la Regla compuesta de Simpson (Algoritmo 4.1, Burden)

##El método.

Sea $n$ un entero par, $h=\dfrac{b-a}{n}$ y $x_j=a+jh$ para $j=0, 1,\dots,$. Entonces:

$$\int_a^bf(x)\,dx=\frac{h}{3}\left[f(a)+2\sum_{j=1}^{(n/2)-1}f(x_{2j})+4\sum_{j=1}^{n/2}f(x_{2j-1})+f(b)\right]$$

con término de error:
 $$-\frac{b-a}{180}\,h^4 f^{(4)}(\mu),\qquad\mu\in(a,b).$$

 $n$ debe ser par porque cada aplicación de la regla de Simpson usa dos subintervalos. Los puntos con índice impar llevan coeficiente 4 y los pares (sin los extremos) coeficiente 2.


```python
def Simpson_compuesta(f, a, b, n):
  if n%2 != 0 or n <= 0:
    raise ValueError("n debe ser un entero positívo par")

  h = (b - a)/n
  XI0 = f(a) + f(b)
  XI1 = 0
  XI2 = 0
  for i in range (1, n):
    X = a +i*h
    if i%2 == 0:
      XI2 += f(X)
    else:
      XI1 += f(X)
  XI = h*(XI0 + XI2*2 + XI1*4)/3
  return XI
```

Se evalúa la función $f(x)= e^{2x}\sin(3x)$ en el intervalo $[0,2]$ con $n=52$ subintervalos

```python
import math

def f(x):
  return math.exp(2*x)*math.sin(3*x)

print(Simpson_compuesta(f, 0, 2, 52))
```
